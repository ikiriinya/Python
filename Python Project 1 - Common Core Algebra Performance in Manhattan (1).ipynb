### 3.5 Extract Borough from DBN (First letter of DBN indicates Borough: M=Manhattan, X=Bronx, K=Brooklyn, Q=Queens, R=Staten Island)

The third character of a school’s DBN identifies its borough. This character was extracted and converted into the corresponding borough name.


```python
df_clean['borough_code'] = df_clean['SCHOOL DBN'].str[2]
borough_map = {'M': 'Manhattan', 'X': 'Bronx', 'K': 'Brooklyn', 'Q': 'Queens', 'R': 'Staten Island'}
df_clean['borough'] = df_clean['borough_code'].map(borough_map)
```

## Performance Comparisons

### Rename the Manhattan mean then join it with Marta Valle’s & Stuyvesant High School data:

```python
manhattan_yearly = (
    algebra_manhattan
    .groupby("YEAR", as_index=False)["MEAN SCORE"]
    .mean()
    .rename(columns={"MEAN SCORE": "Manhattan Mean"})
)

marta_yearly = marta_valle[
    ["YEAR", "MEAN SCORE"]
].rename(columns={"MEAN SCORE": "Marta Valle Mean"}
        
)
Stuyvesant_yearly = Stuyvesant_High[
    ["YEAR", "MEAN SCORE"]
].rename(columns={"MEAN SCORE": "Stuyvesant_High Mean"})
```

## ## 8. Findings and Conclusion

```python
This analysis examined Common Core Algebra performance at Marta Valle High School between 2015 and 2017 and compared 
its results with Stuyvesant High School and the Manhattan school average. Stuyvesant was included as a high-performing
Manhattan comparison school, while the Manhattan average provided a broader borough-level benchmark.

Marta Valle High School’s mean Common Core Algebra score was below the Manhattan average in all three years examined, 
while Stuyvesant High School was consistently above the Manhattan average. In 2015, Marta Valle had a mean score of 61.3, 
compared with 64.96 among Manhattan schools and 82.3 at Stuyvesant. 
Marta Valle’s mean increased to 65.5 in 2016 but declined to 59.9 in 2017. Over the same period, 
the Manhattan average increased from 64.96 to 72.26, while Stuyvesant’s mean increased from 82.3 to 93.9.

The difference between Marta Valle and the Manhattan average increased from 3.66 points in 2015 to 12.36 points in 2017.
The difference between Marta Valle and Stuyvesant also increased, from 21.0 points in 2015 to 34.0 points in 2017. 
Therefore, the results indicate that the performance gaps widened during the period examined.

These findings describe differences in test performance but do not explain their causes. 
The dataset does not include sufficient information about admissions policies, student characteristics, school resources, course placement,
or instructional practices. These factors may contribute to the observed differences.
```
