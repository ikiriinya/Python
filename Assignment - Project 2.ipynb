# Question 3: Create Subset & Standardization

```python
vehicle_subset = df[
    [
        "mpg",
        "hp",
        "wt",
        "cyl"
    ]
].copy()
```

```python
vehicle_subset.head()
```

```python
vehicle_subset= vehicle_subset.rename( 
    columns= {
        "mpg": "miles_per_gallon",
        "hp": "horsepower",
        "wt": "weight",
        "cyl": "cylinders"
    }
)
```

```python
vehicle_subset.head()
```

```python
vehicle_subset.columns.tolist()
```

# Question 4: Aggregate Summary Statistics

```python
vehicle_subset.describe()
```

```python
average_mpg = vehicle_subset [
        "miles_per_gallon"
].mean()

print("average miles_per_gallon:",
     round(average_mpg, 2)
)
```

# Question 5: Segment Analysis 

```python
cylinder_counts = (
    vehicle_subset["cylinders"]
    .value_counts()
    .sort_index()
)

cylinder_counts
```

```python
cylinder_summary = (
    vehicle_subset
    .groupby("cylinders")
    .agg(
        number_of_vehicles=("cylinders", "size"),
        average_mpg=("miles_per_gallon", "mean"),
        average_horsepower=("horsepower", "mean"),
        average_weight=("weight", "mean")
    )
    .round(2)
    .reset_index()
)

cylinder_summary
```

```python
four_cylinder_vehicles = vehicle_subset[
    vehicle_subset["cylinders"] == 4
].copy()

four_cylinder_vehicles
```

```python
four_cylinder_vehicles.describe()
```

```python
ave_mpg_4cyl = (
    four_cylinder_vehicles[
        "miles_per_gallon"
    ]
    .mean()
)

print(
    "Average MPG for four-cylinder vehicles:",
    round(ave_mpg_4cyl, 3)
)
```
