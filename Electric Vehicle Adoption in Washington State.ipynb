```python
top_makes.plot(kind="bar", title="Top 10 Electric Vehicle Manufacturers")
plt.xlabel("Manufacturer")
plt.ylabel("Number of Vehicles")
plt.tight_layout()
```

```python
top_makes.plot(kind="bar", xlabel='Manufacturer', ylabel = "number of vehicles",
title = "Top 10 elextric Vehicle Manufacturer"
)
```

```python
colors = [
    "red", "blue", "green", "orange", "purple",
    "gold", "pink", "teal", "brown", "gray"
]
top_makes.plot(kind="bar", title="Top 10 Electric Vehicle Manufacturers", color = colors)
plt.xlabel("Manufacturer")
plt.ylabel("Number of Vehicles")
plt.tight_layout()
```

```python

top_makes.plot(kind="bar", title="Top 10 Electric Vehicle Manufacturers", cmap='viridis', edgecolor = "red")            
plt.xlabel("Manufacturer")
plt.ylabel("Number of Vehicles")
plt.tight_layout()
```

```python
top_counties = df["County"].value_counts().head(10)
print(top_counties)
```

```python
colors = [
    "red", "blue", "green", "orange", "purple",
    "gold", "pink", "teal", "brown", "gray"
]

top_counties.plot(
    kind="bar",
    color=colors,
    title="Counties with the Most Electric Vehicles"
)

plt.xlabel("County")
plt.ylabel("Number of Vehicles")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
vehicle_types = df["Electric Vehicle Type"].value_counts()
print(vehicle_types)

vehicle_types.plot(
    kind="pie",
    autopct="%1.1f%%",
    title="Electric Vehicle Types"
)
plt.ylabel("")
plt.show()
```
