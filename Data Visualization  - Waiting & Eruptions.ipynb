```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
url = "https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/refs/heads/master/faithful.csv"
```

```python
df = pd.read_csv(url)
```

```python
print(df.head())
```

```python
df.plot(kind='scatter', x='waiting', y='eruptions')
```

```python
plt.show()
```

```python
df.plot(
    kind='scatter',
    x='waiting',
    y='eruptions',
    c='eruptions',                  # Color-code points by eruption duration
    cmap='viridis',                 # Colormap ('viridis', 'plasma', 'magma', 'coolwarm')
)
plt.show()
```

```python
df.plot(
    kind='scatter',
    x='waiting',
    y='eruptions',
    c='eruptions',                  # Color-code points by eruption duration
)
plt.show()
```

```python
df.plot(
    kind='scatter',
    x='waiting',
    y='eruptions',
    c='eruptions',                  # Color-code points by eruption duration
    cmap='viridis',                 # Colormap ('viridis', 'plasma', 'magma', 'coolwarm')
    s=60,                           # Marker size
    alpha=0.7,                      # Transparency (helps spot overlapping points)
    figsize=(10, 6),                # Canvas width and height in inches
    title='Old Faithful: Eruption Duration vs. Waiting Time',
    xlabel='Waiting Time to Next Eruption (minutes)',
    ylabel='Eruption Duration (minutes)',
    grid=True                       # Add background grid lines
)
plt.show()
```

```python
df.plot(
    kind='scatter',
    x='waiting',
    y='eruptions',
    color='blue',                  # Solid point color
    edgecolor='yellow',              # Outline around each point
    linewidth=0.8,
    marker='^',                     # Marker shape ('^' triangle, 's' square, '*' star)
    s=70,
    figsize=(9, 5.5),
    title='Geyser Activity Clusters',
    grid=True
)
plt.show()
```

```python
df.plot(
    kind='scatter',
    x='waiting',
    y='eruptions',
    c='waiting',
    cmap='plasma',
    s=df['waiting'] * 0.8,          # Point size scales with waiting time
    alpha=0.6,
    figsize=(10, 6),
    title='Old Faithful Data (Size & Color Scaled by Waiting Time)',
    grid=True
)
plt.show()
```
